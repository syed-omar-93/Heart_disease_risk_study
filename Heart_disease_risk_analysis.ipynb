```python
# Replace missing values with the calculated median in the cigarettes per day column
patients_at_risk.loc[:, 'Cigs_per_day'] = patients_at_risk['Cigs_per_day'].fillna(at_risk_cigs_per_day_median)
patients_not_at_risk.loc[:, 'Cigs_per_day'] = patients_not_at_risk['Cigs_per_day'].fillna(not_at_risk_cigs_per_day_median)
```

```python
#Replacing the missing values in the BP medication column
patients_at_risk.loc[:, 'On_BP_medication'] = patients_at_risk['On_BP_medication'].fillna(0)
patients_not_at_risk.loc[:, 'On_BP_medication'] = patients_not_at_risk['On_BP_medication'].fillna(0)
```

```python
#Replacing the missing values in the Total cholestrol level column with the median value
at_risk_cholestrol_median = patients_at_risk['Total_cholestrol_level'].median()
not_at_risk_cholestrol_median = patients_not_at_risk['Total_cholestrol_level'].median()

patients_at_risk.loc[:,'Total_cholestrol_level'] = patients_at_risk['Total_cholestrol_level'].fillna(at_risk_cholestrol_median)
patients_not_at_risk.loc[:,'Total_cholestrol_level'] = patients_not_at_risk['Total_cholestrol_level'].fillna(not_at_risk_cholestrol_median)
```

```python
#Impute the missing values of BMI with its median value
at_risk_BMI_median = patients_at_risk['BMI'].median()
not_at_risk_BMI_median = patients_not_at_risk['BMI'].median()

patients_at_risk.loc[:,'BMI'] = patients_at_risk['BMI'].fillna(at_risk_BMI_median)
patients_not_at_risk.loc[:,'BMI'] = patients_not_at_risk['BMI'].fillna(not_at_risk_BMI_median)
```

```python
# Calculating the median heart rate from the training set
at_risk_heart_rate_median = patients_at_risk['Heart_Rate'].median()
not_at_risk_heart_rate_median = patients_not_at_risk['Heart_Rate'].median()


# Now fill the missing values using this median
patients_at_risk.loc[:,'Heart_Rate'] = patients_at_risk['Heart_Rate'].fillna(at_risk_heart_rate_median)
patients_not_at_risk.loc[:,'Heart_Rate'] = patients_not_at_risk['Heart_Rate'].fillna(not_at_risk_heart_rate_median)
```

```python
# Calculate the medians for groups with diabetes and without diabetes
at_risk_diabetes_median = patients_at_risk.loc[patients_at_risk['diabetes_or_not'] == 1, 'Glucose_level'].median()
at_risk_no_diabetes_median = patients_at_risk.loc[patients_at_risk['diabetes_or_not'] == 0, 'Glucose_level'].median()
```

```python
print("The median glucose level of those patients with diabetes is", at_risk_diabetes_median, "mg/dL")
print("The median glucose level of those patients without diabetes is", at_risk_no_diabetes_median, "mg/dL")
```

```python
def impute_glucose_risk(values):

    Glucose_level = values['Glucose_level']
    diabetes_or_not = values['diabetes_or_not']

    if pd.isnull(Glucose_level):
        if diabetes_or_not == 1:
            return at_risk_diabetes_median
        else:
            return at_risk_no_diabetes_median
    else:
        return Glucose_level
```

```python
patients_at_risk.loc[:,'Glucose_level'] = patients_at_risk[['Glucose_level', 'diabetes_or_not']].apply(impute_glucose_risk, axis=1)
```

```python
# Calculate the medians for groups with diabetes and without diabetes
not_at_risk_diabetes_median = patients_not_at_risk.loc[df['diabetes_or_not'] == 1, 'Glucose_level'].median()
not_at_risk_no_diabetes_median = patients_not_at_risk.loc[df['diabetes_or_not'] == 0, 'Glucose_level'].median()
```

```python
def impute_glucose_no_risk(values):

    Glucose_level = values['Glucose_level']
    diabetes_or_not = values['diabetes_or_not']

    if pd.isnull(Glucose_level):
        if diabetes_or_not == 1:
            return not_at_risk_diabetes_median
        else:
            return not_at_risk_no_diabetes_median
    else:
        return Glucose_level
```

```python
patients_not_at_risk.loc[:,'Glucose_level'] = patients_not_at_risk[['Glucose_level', 'diabetes_or_not']].apply(impute_glucose_no_risk, axis=1)
```

```python
#Statistical test to determine if the mean cholestrol level
#of at risk patients is significantly higher than for not at risk patients

#Two sample t-test in python

#Goal: You want to test whether the mean cholesterol level of at-risk patients
#is significantly higher than that of not-at-risk patients
#— which makes it one-tailed.

#To determine whether the mean cholesterol level of at-risk patients is
#significantly higher than that of not-at-risk patients, you should use a
#one-tailed independent samples t-test (also called an independent
#t-test for means)

#We are performing a two-sample t-test — a test used to compare the
#means of two independent groups.
```

```python
#One tailed Welch’s t-test, which is robust to unequal variances and sample sizes.

#Null Hypothesis: The mean cholestrol level of at risk patients
#is less than or equal to the mean cholestrol level of not at risk patients

#Alternative Hypothesis: The mean cholestrol level of at risk patients
#is greater than the mean cholestrol level of not at risk patients


from scipy import stats
from scipy.stats import ttest_ind

alpha = 0.05 #Significance level

# Extract cholestrol levels for each group
at_risk_chol_level = patients_at_risk['Total_cholestrol_level']
not_at_risk_chol_level = patients_not_at_risk['Total_cholestrol_level']

# Perform one-tailed t-test
t_stat, p_value_two_tailed = ttest_ind(at_risk_chol_level, not_at_risk_chol_level ,equal_var=False)

# One-tailed p-value (testing if at-risk > not-at-risk)
p_value_one_tailed = p_value_two_tailed / 2

# Display results
print(f"T-statistic: {t_stat:.3f}")
print(f"One-tailed p-value: {p_value_one_tailed:.3f}")
```

```python
#Since the p-value of 0.000 < 0.05, we reject the null hypothesis and conclude that
#there is statistical evidence to suggest that at risk patients have a higher mean cholesterol level
#than not at risk patients
#So the difference between the mean cholestrol levels of both the groups is statistically significant
```
